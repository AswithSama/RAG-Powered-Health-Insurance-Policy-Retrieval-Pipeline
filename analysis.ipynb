```python
import json
from pathlib import Path


JSON_PATH = Path("policy_chunks.json")


with JSON_PATH.open("r", encoding="utf-8") as file:
    data = json.load(file)


chunks = data["chunks"]

print(f"Total chunks: {len(chunks)}")
```

```python
for chunk in chunks[:3]:
    print("\n" + "=" * 100)
    print(f"Chunk ID: {chunk['chunk_id']}")
    print(f"Chapter ID: {chunk['chapter_id']}")
    print(f"Chapter: {chunk['chapter']}")
    print(f"Token count: {chunk['token_count']}")
    print(f"Headings: {chunk['headings']}")
    print(f"Pages: {chunk['start_printed_page']}–{chunk['end_printed_page']}")
    print("=" * 100)

    print(chunk["content"])
```
